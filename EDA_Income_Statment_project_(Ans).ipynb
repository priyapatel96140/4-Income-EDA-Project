```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('3income.csv')
df
```

```python
#1. Look at the first 5 rows of your data to see what it looks like
df.head(5)
```

```python
#2. Look at the last 5 rows of your data to see what it looks like
df.tail(5)
```

```python
#3. Find Shape of Dataset
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
#4.Check Null Values or hidden missing values(?)
df.isnull().sum()
df=df.replace('?',np.nan)
df.isnull().sum()
```

```python
#5. Drop the missing rows and values:
# Drop any row that contains a missing value
df = df.dropna()

# Check the count again
df.isnull().sum()
```

```python
#6. Dataset Information
df.info()
```

```python
#Univariate Analysis (Analyzing 1 Column at a Time)--
#7 Target(income) Variable Distribution
sns.countplot(x='income', data=df)
plt.title('Distribution of Income Category')
plt.show()
```

```python
#8 Age Distribution
df['age'].hist()
plt.xlabel('age')
plt.ylabel('number of people')
plt.title('Age Distribution')
plt.show()
```

```python
#9 Distribution of Workclass
df['workclass'].value_counts()
df['workclass'].value_counts().plot(kind='bar')
```

```python
#10.Gender distribution in dataset?
print(df['gender'].value_counts())
plt.figure(figsize=(5,5))
df['gender'].value_counts().plot(kind='pie',autopct='%1.1f%%')
plt.title('gender distribution')
plt.show()
```

```python
#Bivariate Analysis (Two Columns)
#11. Which education level has the highest average salary?(Education vs Income) 
df.groupby('education')['income'].mean().sort_values(ascending=False).head(1)
```

```python
#12.Which Workclass Getting Highest Salary
df.groupby('workclass')['income'].mean().sort_values(ascending=False).head(1)
```

```python
#13. Feature Correlation
correlations = df.corr(numeric_only=True)
sns.heatmap(correlations, annot=True)
plt.show()
```
