```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as pl
import seaborn as sns
```

```python
data=pd.read_csv('3income.csv')
data
```

```python
# 1. Display Top 10 rows of the dataset
data.head(10)
```

```python
#2. Display Bottom 10 rows of dataset
data.tail(10)
```

```python
#3. Find Shape of Dataset
print("Rows:", data.shape[0])
print("Columns:", data.shape[1])
```

```python
#4. Dataset Information
data.info()
```

```python
#5. Fetch Random Sample (50%)
data.sample(frac=0.5)
```

```python
#6. Check Null Values
data.isnull().sum()
```

```python
#7. Replace '?' with NaN
data.replace('?', np.nan, inplace=True)
```

```python
#Drop Missing Values
data.dropna(inplace=True)
```

```python
#9. Check Duplicate Data & Drop Them
# Check duplicates
print(data.duplicated().sum())

# Drop duplicates
data.drop_duplicates(inplace=True)
```

```python
#10. Overall Statistics
data.describe(include='all')
```

```python
#Univariate Analysis
#11. Distribution of Age Column
data['age'].hist()
```

```python
#12. Total Persons Age Between 17 to 48
count = data['age'].between(17, 48).sum()

print("Total Persons:", count)
```

```python
#13. Distribution of Workclass
data['workclass'].value_counts()
```

```python
data['workclass'].value_counts().plot(kind='bar')
```

```python
#14. Persons Having Bachelors & Masters Degree
count = data[data['education'].isin(['Bachelors', 'Masters'])].shape[0]

print("Total Persons:", count)
```

```python
#Bivariate Analysis
#15. Replace Salary Values with 0 & 1
data['income'] = data['income'].replace({'<=50K':0, '>50K':1})

print(data['income'].head())
```

```python
#16. Which Workclass Getting Highest Salary
data.groupby('workclass')['income'].mean().sort_values(ascending=False)
```

```python
#17. Male or Female Who Has Better Chance to Get Salary >50K male or female?
data.groupby('gender')['income'].mean().sort_values(ascending=False)
```

```python
#18. Convert Workclass Datatype to Category
data['workclass'] = data['workclass'].astype('category')

print(data['workclass'].dtype)
```

```python
#19. Which Education Level Has Highest Average Salary?
data.groupby('education')['income'].mean().sort_values(ascending=False)
```

```python
#20.Which Age Group Has More People Earning >50K?
high_salary = data[data['income'] == 1]

high_salary['age'].plot(kind='hist')
```
